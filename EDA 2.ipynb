```python
import numpy as np

a = np.arange(1, 21)
print(a)
print(a.ndim, a.shape, a.size, a.dtype)
```

```python
import numpy as np

np.random.seed(0)
a = np.random.randint(0, 51, (4, 5))
print(a)
print(a.itemsize, a.nbytes)
```

```python
import numpy as np

a = np.arange(1, 13).reshape(3, 4)

print(a[1, 2])   
print(a[-1])     
print(a[:, -1])  
```

```python
import numpy as np

a = np.arange(1, 13).reshape(3, 4)

print(a[:2, :2])  
print(a[:, ::2])  
print(a[:, ::-1]) 
```

```python
import numpy as np

a = np.arange(12).reshape(3,4)

v = a[:2, :2]
v[0,0] = 99
print(a)

c = a[:2, :2].copy()
c[0,0] = 100
print(a)
```

```python
import numpy as np

x = np.arange(12)

print(x.reshape(3,4))

a = x.reshape(12,1)
b = x[:, np.newaxis]

print(np.array_equal(a,b))
```

```python
import numpy as np

a = np.random.randint(10, size=(2,3))
b = np.random.randint(10, size=(2,3))

print(np.vstack((a,b)))
print(np.concatenate((a,b), axis=0))

print(np.hstack((a,b)))
print(np.concatenate((a,b), axis=1))
```
