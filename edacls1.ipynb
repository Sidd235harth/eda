```python
#14.	Create two 2x3 arrays of random integers. Concatenate them (a) vertically using np.vstack() and np.concatenate(axis=0), and (b) horizontally using np.hstack() and np.concatenate(axis=1). Confirm both methods give the same result.
import numpy as np
np.random.seed(0)
arr1=np.random.randint(0,21,(2,3))
arr2=np.random.randint(0,21,(2,3))
print(arr1)
print(arr2)
#hor
horizontal=np.hstack([arr1,arr2])
h_concat=np.concatenate([arr1,arr2],axis=1)
print(h_concat)
print(horizontal)
vertical=np.vstack([arr1,arr2])
print(vertical)
```

```python
#Structured Arrays
student_dtype=np.dtype([('name','U10'),('sem','i4'),('cgpa','f8')])
students=np.zeros(3,dtype=student_dtype)

students['name']=['sid','raj','ram']
students['sem']=[5,5,5]
students['cgpa']=[9.8,8.8,9.9]
print(students)
print("names of the students:",students['name'])
print("cgpa of the students : ",students['cgpa'])
print(students[2])
print("student with cgpa less than 9.0 : ",students[students['cgpa']<9.0]['name'])
student_record=students.view(np.recarray)
print(student_record.cgpa)
```
